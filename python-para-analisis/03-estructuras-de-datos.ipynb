La lógica habitual dentro de los default values es:
'''
if key in some_dict:
    value = some_dict[key]
else:
    value = default_value

Los métodos de diccionario como get y pop nos dan un default value de vuelta. Por lo tanto, el bloque anterior de comando se puede escribir como:
value = some_dict.get(key, default_value)

```python
#De manera estándar, get devolverá None si la key no está presente, mientras pop dará una excepción
#Al establecer valor, puede que estos en un diccionario tengan otro tipo de colección, como una lista
#Por ejemplo podemos categorizar una lista de palabras por su primera letra
words = ['apple','bar','bat', 'atom', 'book']
by_letter = {}
type(by_letter)

for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)

by_letter
```

```python
#El método setdefault nos facilita el anterior código, simplemente usando el for
words = ['apple','bar','bat', 'atom', 'book']
by_letter = {}
for word in words:
    letter = word[0]
    by_letter.setdefault(letter,[]).append(word)

by_letter
```

```python
#El módulo collections incorporado tiene una clase muy útil, defaultdicto
#lo cual hace lo anterior mucho más fácil.
#Para crear uno, pasamos un tipo o función que genere el valor defecto para cada slot en el diccionario
from collections import defaultdict

words = ['apple','bar','bat', 'atom', 'book']
by_leter = defaultdict(list)
for word in words:
    by_leter[word[0]].append(word)

by_letter
```

# LIST, SET AND DICTIONARY COMPREHENSIONS
La comprensión de listas es una de las características de Python cómo y muy utilizada
Puedes crear una lista filtrada por elementos de una colección, transformando esos elementos filtrados, podemos trabajar con ellos como queramos. 
La forma habitual de hacerla es:
[expr for value in collection if condition]
Traducido al español
[Haz esto para estos valores en esta colección si cumplen esta condición]
Veamos un ejemplo

```python
#Supongamos el siguiente diccionario
strings = ['a', 'as', 'bat','car', 'dove', 'python']
[x.upper() for x in strings if len(x)>3]
```

```python
#¿Y con diccionarios?
frutas = {'Manzana':1, 'Plátanos':2, 'Peras': 4, 'Kiwi':2}
frutas_escasas = {key: value for key,value in frutas.items() if value <= 2}
frutas_escasas
```

```python
#De la misma manera lo podemos hacer con un set
set_numeros = {1,2,3,4,5}
set_compacto = {x for x in set_numeros if x <= 2}
set_compacto
```

La comprensión de conjuntos y de diccionarios es más conveniente dada su facilidad de escribirlos y de leerlos.
Supongamos que queremos un conjunto que contenga la longitudes de los strings contenidos en una colección, podemos hacerlo mediante la comprensión de set

```python
strings = ['a', 'as', 'bat','car', 'dove', 'python']
longitudes_únicas = {len(x) for x in strings}
longitudes_únicas
```

```python
#También podemos hacerlo mediante la función map
set(map(len, strings))
```

```python
#Mediante la dictionary comprehension podemos crear un mapa de consulta de los strings para su localización
loc_mapping = {value: index for index,value in enumerate(strings)}
loc_mapping
```

# Nested list comprehensions (anidación)
Supongamos que tenemos una lista de nombres en español y en inglés

```python
all_data = [
    ['John', 'Emily', 'Michael', 'Mary', 'Steven'],
    ['Maria', 'Juan', 'Javier', 'Natalia', 'Pilar']
    ] 
#Supongamos que queremos los nombres con dos o más a
result = [name for names in all_data for name in names if name.count("a") >= 2]
result
```

Como hemos visto se hace una doble iteración para obtener el resultado querido

```python
#Veamos un ejempolo con una lita de tuplas
some_tuples = [(1,2,3),(4,5,6),(7,8,9)]
#Supongamos que queremos 'flaten'ear la lista de tuplas
flattened = [x for tuples in some_tuples for x in tuples]
flattened
```

En el caso de que exitan más de dos o tres niveles de anidamiento podemos empezar a plantear hacer una comprensión de lista dentro de otra comprensión de listas
Veamos un ejemplo

```python
equipos = [
    ["Ana", "Luis"],
    ["Marta", "Carlos", "Sofía"],
    ["Pedro", "Elena"]
]
mayúsculas = [[nombre.upper() for nombre in grupos] for grupos in equipos]
mayúsculas
```
