```python
sum = 0
for i in range(101):
    if i%2 == 0:
        sum += i

print(sum)
        
    
```

```python
#Una forma mejor de hacerlo
sum(i for i in range(1,101) if i%2 == 0)

#Todo lo que se pueda simplificar en una línea va a quedar mejor
#y más legible
```
